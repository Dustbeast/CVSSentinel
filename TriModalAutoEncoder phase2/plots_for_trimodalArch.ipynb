```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def plot_reconstruction_errors(diagnostics_path="reconstruction_diagnostics_v4.pt",
                                save_path="fig_recon_errors.pdf"):
    import torch
    import numpy as np

    diag = torch.load(diagnostics_path).numpy()   # [N, 10]
    str_error = diag[:, :9]    # per-metric CVSS error
    mse_sem   = diag[:, 9]     # scalar semantic error

    metric_names = ["AV", "AC", "PR", "UI", "S", "C", "I", "A", "Score"]

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # Left — per-CVSS-metric mean reconstruction error (bar chart)
    mean_errors = str_error.mean(axis=0)
    colors = ["#DC2626" if e > mean_errors.mean() else "#2563EB"
              for e in mean_errors]
    axes[0].bar(metric_names, mean_errors, color=colors, edgecolor="white", linewidth=0.5)
    axes[0].axhline(mean_errors.mean(), color="black", linestyle="--",
                    linewidth=1.0, label="Mean across metrics")
    axes[0].set_title("Mean Reconstruction Error per CVSS Metric", fontweight="bold")
    axes[0].set_xlabel("CVSS Metric")
    axes[0].set_ylabel("Mean Absolute Error")
    axes[0].legend(fontsize=9)

    # Right — semantic MSE distribution (log-scale histogram)
    axes[1].hist(mse_sem, bins=100, color="#7C3AED", edgecolor="none", alpha=0.8)
    axes[1].axvline(np.percentile(mse_sem, 95), color="#DC2626", linestyle="--",
                    linewidth=1.2, label="95th percentile")
    axes[1].set_title("Semantic Reconstruction Error Distribution", fontweight="bold")
    axes[1].set_xlabel("MSE (semantic)")
    axes[1].set_ylabel("Count")
    axes[1].set_yscale("log")
    axes[1].legend(fontsize=9)

    fig.suptitle(
        "Reconstruction Error Analysis — Triple-Projection Fusion AE",
        fontsize=13, fontweight="bold", y=1.01
    )
    fig.tight_layout()
    fig.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.close()
    print(f"Saved: {save_path}")

plot_reconstruction_errors()
```

```python
def plot_umap_features(features_path="node_features_v4.pt",
                       cwe_edges_path="edges_cve_cwe.csv",
                       mapping_path="cve_id_mapping_v4.csv",
                       save_path="fig_umap_features.pdf",
                       n_sample=10000):
    import torch
    import numpy as np
    import pandas as pd
    import umap

    features = torch.load(features_path).numpy()   # [N, 256]
    mapping  = pd.read_csv(mapping_path)
    edges    = pd.read_csv(cwe_edges_path)

    FAKE_CWES = {"NVD-CWE-noinfo", "NVD-CWE-Other"}
    edges     = edges[~edges["Weakness_CWE"].isin(FAKE_CWES)]

    # Get dominant CWE per CVE (most common assignment)
    top_cwes = (
        edges.groupby("Weakness_CWE")["CVE_ID"]
        .count()
        .nlargest(8)
        .index.tolist()
    )

    cve_cwe = edges[edges["Weakness_CWE"].isin(top_cwes)].drop_duplicates("CVE_ID")
    mapping = mapping.merge(cve_cwe, on="CVE_ID", how="left")
    mapping["CWE_label"] = mapping["Weakness_CWE"].fillna("Other")

    # Sample for speed
    np.random.seed(42)
    idx      = np.random.choice(len(features), size=min(n_sample, len(features)),
                                replace=False)
    X_sample = features[idx]
    labels   = mapping["CWE_label"].iloc[idx].values

    # UMAP
    reducer   = umap.UMAP(n_components=2, random_state=42, n_neighbors=30, min_dist=0.1)
    embedding = reducer.fit_transform(X_sample)

    # Plot
    unique_labels = sorted(set(labels))
    cmap = plt.cm.get_cmap("tab10", len(unique_labels))
    color_map = {l: cmap(i) for i, l in enumerate(unique_labels)}

    fig, ax = plt.subplots(figsize=(12, 9))
    for label in unique_labels:
        mask = labels == label
        ax.scatter(
            embedding[mask, 0], embedding[mask, 1],
            c=[color_map[label]], label=label,
            s=4, alpha=0.5, linewidths=0
        )

    ax.set_title(
        f"UMAP Projection of Fused CVE Embeddings (n={n_sample:,})\n"
        "Colored by dominant CWE category",
        fontsize=13, fontweight="bold"
    )
    ax.set_xlabel("UMAP-1")
    ax.set_ylabel("UMAP-2")
    ax.legend(markerscale=3, fontsize=9, loc="best",
              framealpha=0.8, title="CWE Category")
    ax.set_xticks([])
    ax.set_yticks([])

    fig.tight_layout()
    fig.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.close()
    print(f"Saved: {save_path}")

plot_umap_features()
```

```python
def plot_anomaly_scores(dominant_scores, cola_scores, anomalydae_scores,
                        mapping_path="cve_id_mapping_v4.csv",
                        top_k=50,
                        save_path="fig_anomaly_scores.pdf"):
    import numpy as np
    import pandas as pd
    from scipy.stats import rankdata

    mapping = pd.read_csv(mapping_path)

    def rank_normalize(s):
        return rankdata(s) / len(s)

    final_score = (
        rank_normalize(dominant_scores) +
        rank_normalize(cola_scores) +
        rank_normalize(anomalydae_scores)
    ) / 3

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # Left — score distribution with tail threshold
    threshold = np.percentile(final_score, 99)
    axes[0].hist(final_score, bins=200, color="#2563EB", edgecolor="none", alpha=0.8)
    axes[0].axvline(threshold, color="#DC2626", linestyle="--", linewidth=1.5,
                    label=f"99th percentile = {threshold:.3f}")
    axes[0].set_title("Ensemble Anomaly Score Distribution", fontweight="bold")
    axes[0].set_xlabel("Percentile Rank Score")
    axes[0].set_ylabel("Number of CVEs")
    axes[0].set_yscale("log")
    axes[0].legend(fontsize=10)

    # Right — top-K CVEs bar chart
    top_idx  = np.argsort(final_score)[::-1][:top_k]
    top_cves = mapping.iloc[top_idx]["CVE_ID"].values
    top_vals = final_score[top_idx]

    axes[1].barh(
        range(top_k), top_vals[::-1],
        color="#DC2626", edgecolor="none", alpha=0.85
    )
    axes[1].set_yticks(range(top_k))
    axes[1].set_yticklabels(top_cves[::-1], fontsize=6)
    axes[1].set_xlabel("Ensemble Anomaly Score")
    axes[1].set_title(f"Top {top_k} Flagged CVEs", fontweight="bold")
    axes[1].axvline(threshold, color="#2563EB", linestyle="--",
                    linewidth=1.0, label="99th percentile")
    axes[1].legend(fontsize=9)

    fig.suptitle(
        "CVSS Inconsistency Detection — Final Anomaly Scores",
        fontsize=13, fontweight="bold", y=1.01
    )
    fig.tight_layout()
    fig.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.close()
    print(f"Saved: {save_path}")
```

```python
def plot_roc_curves(dominant_scores, cola_scores, anomalydae_scores,
                    ground_truth_labels,
                    save_path="fig_roc_curves.pdf"):
    from sklearn.metrics import roc_curve, auc, average_precision_score
    import numpy as np
    from scipy.stats import rankdata

    def rank_normalize(s):
        return rankdata(s) / len(s)

    ensemble = (
        rank_normalize(dominant_scores) +
        rank_normalize(cola_scores) +
        rank_normalize(anomalydae_scores)
    ) / 3

    models = {
        "DOMINANT":    dominant_scores,
        "CoLA":        cola_scores,
        "AnomalyDAE":  anomalydae_scores,
        "Ensemble":    ensemble,
    }
    colors = {
        "DOMINANT":   "#2563EB",
        "CoLA":       "#7C3AED",
        "AnomalyDAE": "#F59E0B",
        "Ensemble":   "#DC2626",
    }

    fig, axes = plt.subplots(1, 2, figsize=(14, 6))

    # Left — ROC curves
    for name, scores in models.items():
        fpr, tpr, _ = roc_curve(ground_truth_labels, scores)
        roc_auc     = auc(fpr, tpr)
        axes[0].plot(fpr, tpr, color=colors[name], linewidth=2.0,
                     label=f"{name}  (AUC = {roc_auc:.3f})")

    axes[0].plot([0, 1], [0, 1], "k--", linewidth=0.8, label="Random")
    axes[0].set_title("ROC Curves — Synthetic Anomaly Detection", fontweight="bold")
    axes[0].set_xlabel("False Positive Rate")
    axes[0].set_ylabel("True Positive Rate")
    axes[0].legend(fontsize=10)
    axes[0].set_xlim([0, 1])
    axes[0].set_ylim([0, 1.02])

    # Right — Precision-Recall curves (more informative for imbalanced data)
    for name, scores in models.items():
        from sklearn.metrics import precision_recall_curve
        prec, rec, _ = precision_recall_curve(ground_truth_labels, scores)
        ap = average_precision_score(ground_truth_labels, scores)
        axes[1].plot(rec, prec, color=colors[name], linewidth=2.0,
                     label=f"{name}  (AP = {ap:.3f})")

    baseline = ground_truth_labels.mean()
    axes[1].axhline(baseline, color="black", linestyle="--",
                    linewidth=0.8, label=f"Random (AP = {baseline:.3f})")
    axes[1].set_title("Precision-Recall Curves", fontweight="bold")
    axes[1].set_xlabel("Recall")
    axes[1].set_ylabel("Precision")
    axes[1].legend(fontsize=10)
    axes[1].set_xlim([0, 1])
    axes[1].set_ylim([0, 1.02])

    fig.suptitle(
        "Model Evaluation on Synthetic CVSS Anomalies",
        fontsize=13, fontweight="bold", y=1.01
    )
    fig.tight_layout()
    fig.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.close()
    print(f"Saved: {save_path}")
```
